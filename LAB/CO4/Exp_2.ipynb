```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)
rng = np.random.default_rng(42)

L = np.array([[0.8,0.1],[0.7,0.2],[0.2,0.8],[0.1,0.9]])   # Λ (4x2)
psi = 1 - (L**2).sum(axis=1)                              # unique variances
Psi = np.diag(psi)                                        # Ψ (diagonal)

# Covariance implied by the model: Cov(X) = ΛΛᵀ + Ψ
Sigma = L @ L.T + Psi
print("Cov(X) = LL' + Psi:\n", Sigma)

# Simulate customers: X = ΛF + ε,  F ~ N(0,I),  ε ~ N(0,Ψ)
n = 2000
F = rng.standard_normal((n, 2))                           # latent factors
E = rng.standard_normal((n, 4)) * np.sqrt(psi)            # unique errors
X = F @ L.T + E                                           # observed data (n x 4)
print("Shape of X:", X.shape)
print("Sample covariance:\n", np.cov(X.T))
```
